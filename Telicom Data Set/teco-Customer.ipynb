### 27. Find the churn rate (%) for each contract type.

```python
# 1st We create the Churn_Num column Where Yes = 1, No = 0
df['Churn_Num'] = df['Churn'].map({'Yes': 1, 'No':0})

# . Group by Contract and calculate the mean and multiply by 100
churn_rate_contract = round(df.groupby('Contract')['Churn_Num'].mean() * 100, 2)
print(churn_rate_contract)
```

```python
%%sql
SELECT
    Contract,
    ROUND(sum(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) as Churn_Rate_Percentage
FROM churn_table
    GROUP BY 
        Contract;

```

### 30. Which combination of InternetService and Contract has the highest churn count?

```python
# 1. Filter for churned customers first
churned_df = df[df['Churn'] == 'Yes']

# 2. Group by BOTH columns and use .size() to count the rows
# .reset_index() just makes the output look like a clean dataframe!
combo_counts = churned_df.groupby(['InternetService', 'Contract']).size().reset_index(name='ChurnCount')

# 3. Sort to see the highest at the top
result = combo_counts.sort_values(by='ChurnCount', ascending=False)
print(result)
```

```python
%%sql
SELECT 
    InternetService, 
    Contract, 
    COUNT(*) AS ChurnCount
FROM 
    churn_table
WHERE 
    Churn = 'Yes'
GROUP BY 
    InternetService, 
    Contract
ORDER BY 
    ChurnCount DESC;
```
